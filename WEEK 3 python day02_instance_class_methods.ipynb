```python
# day02_instance_class_methods.py

class Employee:

    # Class Variables
    company_name = "ABC Technologies"
    company_location = "Hyderabad"
    employee_count = 0

    def __init__(self, employee_id, name, department, salary, email):
        self.employee_id = employee_id
        self.name = name
        self.department = department
        self.email = email

        # Private-style attribute
        self.__salary = 0
        self.set_salary(salary)

        # Increase employee count
        Employee.employee_count += 1

    # Getter Method
    def get_salary(self):
        return self.__salary

    # Setter Method with Validation
    def set_salary(self, salary):
        if self.validate_salary(salary):
            self.__salary = salary
        else:
            print("Invalid Salary. Salary must be greater than 0.")

    # Instance Method
    def validate_salary(self, salary):
        return salary > 0

    # Instance Method
    def calculate_bonus(self):
        return self.__salary * 0.10

    # Instance Method
    def display_employee(self):
        print("\nEmployee Details")
        print("----------------")
        print("ID:", self.employee_id)
        print("Name:", self.name)
        print("Department:", self.department)
        print("Email:", self.email)
        print("Salary:", self.__salary)
        print("Bonus:", self.calculate_bonus())

    # Class Method
    @classmethod
    def get_employee_count(cls):
        return cls.employee_count

    # Class Method
    @classmethod
    def get_company_details(cls):
        print("\nCompany Details")
        print("----------------")
        print("Company Name:", cls.company_name)
        print("Location:", cls.company_location)
        print("Employee Count:", cls.employee_count)

    # Alternative Constructor
    @classmethod
    def from_string(cls, emp_string):
        emp_id, name, department, salary = emp_string.split(",")
        email = name.lower() + "@company.com"
        return cls(int(emp_id), name, department, float(salary), email)

    # Static Method
    @staticmethod
    def is_valid_email(email):
        return "@" in email and "." in email

    # Static Method
    @staticmethod
    def is_valid_salary(salary):
        return salary > 0


# ------------------------------
# TEST CASES
# ------------------------------

emp1 = Employee(
    101,
    "John",
    "Python",
    50000,
    "john@gmail.com"
)

emp2 = Employee(
    102,
    "Ajith",
    "Data Science",
    60000,
    "ajith@gmail.com"
)

emp3 = Employee.from_string(
    "103,Rahul,Java,55000"
)

# Display Employees
emp1.display_employee()
emp2.display_employee()
emp3.display_employee()

# Company Details
Employee.get_company_details()

# Employee Count
print("\nTotal Employees:",
      Employee.get_employee_count())

# Static Method Testing
print("\nEmail Validation:")
print(Employee.is_valid_email("test@gmail.com"))

print("\nSalary Validation:")
print(Employee.is_valid_salary(50000))

# Getter Method
print("\nCurrent Salary:",
      emp1.get_salary())

# Setter Method
emp1.set_salary(70000)

print("Updated Salary:",
      emp1.get_salary())

print("Updated Bonus:",
      emp1.calculate_bonus())
```
