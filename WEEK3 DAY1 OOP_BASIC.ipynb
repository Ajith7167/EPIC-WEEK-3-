```python
# ==========================================
# Day 01 - OOP Basics
# Classes, Objects & Constructors
# ==========================================

# Employee Class
class Employee:

    def __init__(self, employee_id, name, department, salary, experience):
        self.employee_id = employee_id
        self.name = name
        self.department = department
        self.salary = salary
        self.experience = experience

    def display_details(self):
        print("\nEmployee Details")
        print("Employee ID :", self.employee_id)
        print("Name        :", self.name)
        print("Department  :", self.department)
        print("Salary      :", self.salary)
        print("Experience  :", self.experience, "Years")

    def calculate_bonus(self):
        if self.experience >= 5:
            bonus = self.salary * 0.20
        else:
            bonus = self.salary * 0.10

        print("Bonus :", bonus)
        return bonus

    def calculate_annual_salary(self):
        annual_salary = self.salary * 12
        print("Annual Salary :", annual_salary)
        return annual_salary

    def update_salary(self, new_salary):
        if new_salary > 0:
            self.salary = new_salary
            print("Salary Updated Successfully")
        else:
            print("Invalid Salary")


# ==========================================
# Creating Employee Objects
# ==========================================

emp1 = Employee("EMP001", "Ajith", "IT", 50000, 2)
emp2 = Employee("EMP002", "Rahul", "HR", 45000, 6)
emp3 = Employee("EMP003", "Priya", "Finance", 55000, 4)
emp4 = Employee("EMP004", "Kiran", "Testing", 40000, 7)
emp5 = Employee("EMP005", "Divya", "Development", 60000, 5)

employees = [emp1, emp2, emp3, emp4, emp5]

for emp in employees:
    emp.display_details()
    emp.calculate_bonus()
    emp.calculate_annual_salary()
    print("-" * 40)

# Modify Attribute
emp1.update_salary(55000)

print("\nUpdated Details")
emp1.display_details()
```

```python
#Student Class
class Student:

    def __init__(self, student_id, name, marks):
        self.student_id = student_id
        self.name = name
        self.marks = marks

    def display(self):
        print(f"Student ID: {self.student_id}")
        print(f"Name: {self.name}")
        print(f"Marks: {self.marks}")
```

```python
#Bank Account class
class BankAccount:

    def __init__(self, account_number, holder_name, balance):
        self.account_number = account_number
        self.holder_name = holder_name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def display(self):
        print(f"Balance: {self.balance}")
```

```python
#product class
class Product:

    def __init__(self, product_id, product_name, price):
        self.product_id = product_id
        self.product_name = product_name
        self.price = price

    def display(self):
        print(self.product_name, "-", self.price)
```

```python
#car
class Car:

    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price

    def display(self):
        print(self.brand, self.model, self.price)
```

```python
class Book:

    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def display(self):
        print(self.title, "-", self.author)
```

```python
student1 = Student("S101", "Ajith", 85)
student1.display()

account1 = BankAccount("12345", "Ajith", 10000)
account1.deposit(5000)
account1.display()

product1 = Product("P101", "Laptop", 55000)
product1.display()

car1 = Car("Hyundai", "Creta", 1500000)
car1.display()

book1 = Book("Python Basics", "John", 499)
book1.display()
```
