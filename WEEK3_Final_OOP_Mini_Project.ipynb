```python
from abc import ABC, abstractmethod


# Employee Classes

class Employee(ABC):
    company_name = "EPIC Technologies"

    def __init__(self, emp_id, name, department, salary):
        self.__emp_id = emp_id
        self.__name = name
        self.__department = department
        self.__salary = salary

    # Encapsulation (Getters)
    def get_emp_id(self):
        return self.__emp_id

    def get_name(self):
        return self.__name

    def get_department(self):
        return self.__department

    def get_salary(self):
        return self.__salary

    # Encapsulation (Setter)
    def set_salary(self, salary):
        if salary > 0:
            self.__salary = salary

    @classmethod
    def company_details(cls):
        return f"Company: {cls.company_name}"

    @staticmethod
    def validate_salary(salary):
        return salary > 0

    @abstractmethod
    def calculate_bonus(self):
        pass

    def calculate_annual_salary(self):
        return self.get_salary() * 12

    def display_details(self):
        print(
            f"ID: {self.get_emp_id()}, "
            f"Name: {self.get_name()}, "
            f"Department: {self.get_department()}, "
            f"Salary: {self.get_salary()}"
        )
```

```python
# =========================
# Child Classes
# =========================

class Developer(Employee):

    def __init__(self, emp_id, name, department, salary, language):
        super().__init__(emp_id, name, department, salary)
        self.language = language

    def calculate_bonus(self):
        return self.get_salary() * 0.15


class Manager(Employee):

    def __init__(self, emp_id, name, department, salary, team_size):
        super().__init__(emp_id, name, department, salary)
        self.team_size = team_size

    def calculate_bonus(self):
        return self.get_salary() * 0.20
```

```python
class HRManager(Employee):

    def __init__(self, emp_id, name, department, salary, employees_handled):
        super().__init__(emp_id, name, department, salary)
        self.employees_handled = employees_handled

    def calculate_bonus(self):
        return self.get_salary() * 0.10
```

```python

# =========================
# Validation Service
# =========================

class ValidationService:

    @staticmethod
    def validate_name(name):
        return len(name.strip()) > 0

    @staticmethod
    def validate_salary(salary):
        return salary > 0
```

```python
# =========================
# Employee Service
# =========================

class EmployeeService:

    def __init__(self):
        self.employees = []

    def add_employee(self, employee):
        self.employees.append(employee)
        print("Employee Added Successfully")

    def display_employees(self):
        if not self.employees:
            print("No Employees Found")
            return

        for emp in self.employees:
            emp.display_details()

    def search_employee(self, emp_id):
        for emp in self.employees:
            if emp.get_emp_id() == emp_id:
                return emp
        return None

    def update_employee(self, emp_id, new_salary):
        emp = self.search_employee(emp_id)

        if emp:
            emp.set_salary(new_salary)
            print("Employee Updated Successfully")
        else:
            print("Employee Not Found")

    def delete_employee(self, emp_id):
        emp = self.search_employee(emp_id)

        if emp:
            self.employees.remove(emp)
            print("Employee Deleted Successfully")
        else:
            print("Employee Not Found")

    def search_by_department(self, department):
        found = False

        for emp in self.employees:
            if emp.get_department().lower() == department.lower():
                emp.display_details()
                found = True

        if not found:
            print("No Employees Found")
```

```python

# =========================
# Report Service
# =========================

class ReportService:

    def generate_report(self, employees):

        print("\n===== EMPLOYEE REPORT =====")

        for emp in employees:
            print(
                f"{emp.get_name()} | "
                f"Department: {emp.get_department()} | "
                f"Salary: {emp.get_salary()} | "
                f"Bonus: {emp.calculate_bonus()}"
            )
```

```python
# =========================
# Main Program
# =========================

def main():

    employee_service = EmployeeService()
    report_service = ReportService()

    while True:

        print("\n===== Employee Management System V3 =====")
        print("1. Add Employee")
        print("2. Display Employees")
        print("3. Search Employee")
        print("4. Update Employee")
        print("5. Delete Employee")
        print("6. Search By Department")
        print("7. Generate Report")
        print("8. Exit")

        choice = input("Enter Choice: ")

        if choice == "1":

            emp_type = input("Developer/Manager/HRManager: ")
            emp_id = int(input("Enter ID: "))
            name = input("Enter Name: ")
            department = input("Enter Department: ")
            salary = float(input("Enter Salary: "))

            if not ValidationService.validate_name(name):
                print("Invalid Name")
                continue

            if not ValidationService.validate_salary(salary):
                print("Invalid Salary")
                continue

            if emp_type.lower() == "developer":
                language = input("Programming Language: ")
                emp = Developer(
                    emp_id,
                    name,
                    department,
                    salary,
                    language
                )

            elif emp_type.lower() == "manager":
                team_size = int(input("Team Size: "))
                emp = Manager(
                    emp_id,
                    name,
                    department,
                    salary,
                    team_size
                )

            elif emp_type.lower() == "hrmanager":
                handled = int(input("Employees Handled: "))
                emp = HRManager(
                    emp_id,
                    name,
                    department,
                    salary,
                    handled
                )

            else:
                print("Invalid Employee Type")
                continue

            employee_service.add_employee(emp)

        elif choice == "2":
            employee_service.display_employees()

        elif choice == "3":
            emp_id = int(input("Enter Employee ID: "))
            emp = employee_service.search_employee(emp_id)

            if emp:
                emp.display_details()
            else:
                print("Employee Not Found")

        elif choice == "4":
            emp_id = int(input("Enter Employee ID: "))
            salary = float(input("Enter New Salary: "))
            employee_service.update_employee(emp_id, salary)

        elif choice == "5":
            emp_id = int(input("Enter Employee ID: "))
            employee_service.delete_employee(emp_id)

        elif choice == "6":
            department = input("Enter Department: ")
            employee_service.search_by_department(department)

        elif choice == "7":
            report_service.generate_report(
                employee_service.employees
            )

        elif choice == "8":
            print("Thank You")
            break

        else:
            print("Invalid Choice")


main()
```
