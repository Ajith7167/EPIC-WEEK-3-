```python
from abc import ABC, abstractmethod


# -------------------------------
# Abstract Account Class
# -------------------------------

class Account(ABC):

    def __init__(self, account_number, customer, balance=0):
        self.account_number = account_number
        self.customer = customer
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("Invalid deposit amount")
            return

        self.balance += amount
        print(f"₹{amount} deposited successfully")
        print(f"Current Balance: ₹{self.balance}")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")
            return

        if amount > self.balance:
            print("Insufficient balance")
            return

        self.balance -= amount
        print(f"₹{amount} withdrawn successfully")
        print(f"Current Balance: ₹{self.balance}")

    @abstractmethod
    def calculate_interest(self):
        pass

    def display_account_details(self):
        print("\nAccount Details")
        print("------------------------")
        print("Account Number:", self.account_number)
        print("Customer Name:", self.customer.name)
        print("Balance:", self.balance)












```

```python
# -------------------------------
# Customer Class
# -------------------------------

class Customer:

    def __init__(self, customer_id, name, email, phone):
        self.customer_id = customer_id
        self.name = name
        self.email = email
        self.phone = phone

    def display_customer_details(self):
        print("\nCustomer Details")
        print("------------------------")
        print("Customer ID:", self.customer_id)
        print("Name:", self.name)
        print("Email:", self.email)
        print("Phone:", self.phone)
```

```python
# -------------------------------
# Savings Account
# -------------------------------

class SavingsAccount(Account):

    interest_rate = 4

    def calculate_interest(self):

        interest = self.balance * self.interest_rate / 100

        print(f"Savings Interest: ₹{interest}")

        return interest
```

```python
# -------------------------------
# Current Account
# -------------------------------

class CurrentAccount(Account):

    minimum_balance = 1000

    def calculate_interest(self):

        print("Current Account does not provide interest")

        return 0

    def withdraw(self, amount):

        if amount <= 0:
            print("Invalid withdrawal amount")
            return

        if self.balance - amount < self.minimum_balance:
            print(
                f"Withdrawal denied. Minimum balance of "
                f"₹{self.minimum_balance} must be maintained."
            )
            return

        self.balance -= amount

        print(f"₹{amount} withdrawn successfully")
        print(f"Current Balance: ₹{self.balance}")
```

```python


# -------------------------------
# Bank Class
# -------------------------------

class Bank:

    def __init__(self, bank_name):
        self.bank_name = bank_name
        self.customers = []
        self.accounts = []

    def add_customer(self, customer):

        self.customers.append(customer)

        print(
            f"Customer {customer.name} added successfully"
        )

    def add_account(self, account):

        self.accounts.append(account)

        print(
            f"Account {account.account_number} created successfully"
        )

    def display_customers(self):

        print("\nAll Customers")
        print("========================")

        for customer in self.customers:
            customer.display_customer_details()

    def display_accounts(self):

        print("\nAll Accounts")
        print("========================")

        for account in self.accounts:
            account.display_account_details()
```

```python
# ====================================================
# TESTING
# ====================================================


print("\n========== BANKING SYSTEM ==========")


# Create Bank

bank = Bank("ABC Bank")


# -------------------------------
# Customer 1
# -------------------------------

customer1 = Customer(
    101,
    "Ajith Kumar",
    "ajith@gmail.com",
    "9876543210"
)


# -------------------------------
# Customer 2
# -------------------------------

customer2 = Customer(
    102,
    "Rahul Sharma",
    "rahul@gmail.com",
    "9876501234"
)


# Add customers to Bank

bank.add_customer(customer1)

bank.add_customer(customer2)


# -------------------------------
# Create Savings Account
# -------------------------------

savings = SavingsAccount(
    "SAV001",
    customer1,
    10000
)


# -------------------------------
# Create Current Account
# -------------------------------

current = CurrentAccount(
    "CUR001",
    customer2,
    15000
)
# Add accounts to Bank

bank.add_account(savings)

bank.add_account(current)
```
