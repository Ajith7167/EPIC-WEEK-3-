```python
# 1. Common Exception Examples
print("COMMON EXCEPTION EXAMPLES")
# ValueError
try:
    number = int("abc")
except ValueError:
    print("ValueError: Please enter a valid number.")


# TypeError
try:
    result = "10" + 5
except TypeError:
    print("TypeError: Cannot add string and integer.")


# IndexError
try:
    numbers = [10, 20, 30]
    print(numbers[5])
except IndexError:
    print("IndexError: List index is not available.")


# KeyError
try:
    employee = {
        "name": "Ajith",
        "salary": 30000
    }

    print(employee["department"])

except KeyError:
    print("KeyError: Key does not exist.")


# ZeroDivisionError
try:
    result = 10 / 0
except ZeroDivisionError:
    print("ZeroDivisionError: Cannot divide by zero.")


# FileNotFoundError
try:
    file = open("employee.txt", "r")
except FileNotFoundError:
    print("FileNotFoundError: File does not exist.")

# 2. try, except, else, finally
print("\n===== TRY EXCEPT ELSE FINALLY =====")
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))
    result = num1 / num2
except ValueError:
    print("Please enter numbers only.")
except ZeroDivisionError:
    print("Second number cannot be zero.")
else:
    print("Result:", result)
finally:

    print("Calculation completed.")


# ==========================================
# CUSTOM EXCEPTIONS
# ==========================================


class InvalidEmployeeIDError(Exception):
    pass


class InvalidSalaryError(Exception):
    pass


class InvalidEmailError(Exception):
    pass


class InvalidAgeError(Exception):
    pass


# ==========================================
# VALIDATION FUNCTIONS
# ==========================================


def validate_employee_id(employee_id):

    if not employee_id.startswith("EMP"):
        raise InvalidEmployeeIDError(
            "Employee ID must start with EMP."
        )

    if len(employee_id) < 4:
        raise InvalidEmployeeIDError(
            "Employee ID is too short."
        )

    return True


def validate_name(name):

    if len(name.strip()) < 2:
        raise ValueError(
            "Employee name must contain at least 2 characters."
        )

    if not name.replace(" ", "").isalpha():
        raise ValueError(
            "Employee name must contain only letters."
        )

    return True


def validate_age(age):

    if age < 18:
        raise InvalidAgeError(
            "Employee age must be at least 18."
        )

    if age > 60:
        raise InvalidAgeError(
            "Employee age cannot be greater than 60."
        )

    return True


def validate_email(email):

    if "@" not in email or "." not in email:
        raise InvalidEmailError(
            "Please enter a valid email address."
        )

    return True


def validate_salary(salary):

    if salary <= 0:
        raise InvalidSalaryError(
            "Salary must be greater than zero."
        )

    return True


def validate_department(department):

    departments = [
        "IT",
        "HR",
        "Finance",
        "Sales",
        "Marketing"
    ]

    if department not in departments:
        raise ValueError(
            "Department must be IT, HR, Finance, Sales or Marketing."
        )

    return True


# ==========================================
# EMPLOYEE VALIDATION SYSTEM
# ==========================================


print("\n===== EMPLOYEE VALIDATION SYSTEM =====")

try:

    employee_id = input("Enter Employee ID: ")
    validate_employee_id(employee_id)

    name = input("Enter Employee Name: ")
    validate_name(name)

    age = int(input("Enter Age: "))
    validate_age(age)

    email = input("Enter Email: ")
    validate_email(email)

    salary = float(input("Enter Salary: "))
    validate_salary(salary)

    department = input(
        "Enter Department (IT/HR/Finance/Sales/Marketing): "
    )

    validate_department(department)


except InvalidEmployeeIDError as error:

    print("Employee ID Error:", error)


except InvalidAgeError as error:

    print("Age Error:", error)


except InvalidEmailError as error:

    print("Email Error:", error)


except InvalidSalaryError as error:

    print("Salary Error:", error)


except ValueError as error:

    print("Input Error:", error)


else:

    print("\nEmployee registered successfully.")

    print("\nEmployee Details")

    print("Employee ID :", employee_id)
    print("Name        :", name)
    print("Age         :", age)
    print("Email       :", email)
    print("Salary      :", salary)
    print("Department  :", department)


finally:

    print("\nEmployee validation process completed.")
```
