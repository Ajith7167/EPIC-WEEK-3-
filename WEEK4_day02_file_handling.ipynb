```python
import os
import csv
import json

# Create data folder
os.makedirs("data", exist_ok=True)

TXT_FILE = "data/employees.txt"
CSV_FILE = "data/employees.csv"
JSON_FILE = "data/employees.json"



# Validate Employee Data

def validate_employee(employee):

    if employee["id"] == "":
        print("Invalid Employee ID")
        return False

    if employee["name"] == "":
        print("Invalid Employee Name")
        return False

    if employee["department"] == "":
        print("Invalid Department")
        return False

    try:
        salary = float(employee["salary"])

        if salary <= 0:
            print("Invalid Salary")
            return False

    except ValueError:
        print("Salary must be numeric")
        return False

    return True
```

```python
# Load Employees from JSON
def load_employees():

    try:

        with open(JSON_FILE, "r") as file:

            content = file.read()

            if content == "":
                return []

            return json.loads(content)

    except FileNotFoundError:

        print("JSON file not found")
        return []

    except json.JSONDecodeError:

        print("Invalid JSON file")
        return []

```

```python
# Rewrite TXT File
def rewrite_txt(employees):

    with open(TXT_FILE, "w") as file:

        for employee in employees:

            file.write(
                employee["id"] + "," +
                employee["name"] + "," +
                employee["department"] + "," +
                employee["salary"] + "\n"
            )
```

```python
# Rewrite CSV File
def rewrite_csv(employees):

    with open(CSV_FILE, "w", newline="") as file:

        writer = csv.writer(file)

        writer.writerow(
            ["ID", "Name", "Department", "Salary"]
        )

        for employee in employees:

            writer.writerow([
                employee["id"],
                employee["name"],
                employee["department"],
                employee["salary"]
            ])
```

```python
# Save Employee
def save_employee(employee):

    if not validate_employee(employee):
        return

    employees = load_employees()

    for emp in employees:

        if emp["id"] == employee["id"]:
            print("Duplicate Employee ID")
            return

    employees.append(employee)

    with open(JSON_FILE, "w") as file:

        json.dump(employees, file, indent=4)

    rewrite_txt(employees)
    rewrite_csv(employees)

    print(employee["id"], "saved successfully")
```

```python
# Display Employees
def display_employees():

    employees = load_employees()

    if len(employees) == 0:

        print("No Employees Found")
        return

    for employee in employees:

        print("--------------------")
        print("ID:", employee["id"])
        print("Name:", employee["name"])
        print("Department:", employee["department"])
        print("Salary:", employee["salary"])
```

```python
# Search Employee
def search_employee(employee_id):

    employees = load_employees()

    for employee in employees:

        if employee["id"] == employee_id:

            print("\nEmployee Found")
            print(employee)

            return

    print("Employee Not Found")
```

```python
# Update Employee
def update_employee(employee_id, updated_employee):

    employees = load_employees()

    found = False

    for employee in employees:

        if employee["id"] == employee_id:

            employee["name"] = updated_employee["name"]
            employee["department"] = updated_employee["department"]
            employee["salary"] = updated_employee["salary"]

            found = True
            break

    if found:

        with open(JSON_FILE, "w") as file:

            json.dump(employees, file, indent=4)

        rewrite_txt(employees)
        rewrite_csv(employees)

        print("Employee Updated")

    else:

        print("Employee Not Found")
```

```python
# Delete Employee
def delete_employee(employee_id):

    employees = load_employees()

    found = False

    for employee in employees:

        if employee["id"] == employee_id:

            employees.remove(employee)

            found = True
            break

    if found:

        with open(JSON_FILE, "w") as file:

            json.dump(employees, file, indent=4)

        rewrite_txt(employees)
        rewrite_csv(employees)

        print("Employee Deleted")

    else:

        print("Employee Not Found")
```

```python
# Read TXT File
def read_txt():

    try:

        with open(TXT_FILE, "r") as file:

            print("\nTXT FILE DATA")
            print(file.read())

    except FileNotFoundError:

        print("TXT File Not Found")
```

```python
# Read CSV File
def read_csv():

    try:

        with open(CSV_FILE, "r") as file:

            reader = csv.reader(file)

            print("\nCSV FILE DATA")

            for row in reader:

                print(row)

    except FileNotFoundError:

        print("CSV File Not Found")
```

```python
# Test Data
employee1 = {
    "id": "EMP001",
    "name": "Ajith",
    "department": "IT",
    "salary": "50000"
}

employee2 = {
    "id": "EMP002",
    "name": "Rahul",
    "department": "HR",
    "salary": "45000"
}

employee3 = {
    "id": "EMP003",
    "name": "Priya",
    "department": "Finance",
    "salary": "55000"
}


# Add Employees
save_employee(employee1)
save_employee(employee2)
save_employee(employee3)


# Display Employees
print("\nALL EMPLOYEES")
display_employees()


# Search Employee
print("\nSEARCH EMPLOYEE")
search_employee("EMP002")


# Update Employee
updated_employee = {
    "id": "EMP002",
    "name": "Rahul Sharma",
    "department": "HR",
    "salary": "60000"
}

update_employee("EMP002", updated_employee)


# Display After Update
print("\nAFTER UPDATE")
display_employees()


# Delete Employee

delete_employee("EMP003")


# Display After Delete
print("\nAFTER DELETE")
display_employees()


# Read TXT and CSV Files
read_txt()
read_csv()
```
