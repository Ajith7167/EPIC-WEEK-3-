```python
# Parent Class
class Employee:

    def __init__(self, emp_id, name, salary, department):
        self.emp_id = emp_id
        self.name = name
        self._salary = salary          # Protected attribute
        self.department = department

    def display_details(self):
        print("\nEmployee Details")
        print("ID:", self.emp_id)
        print("Name:", self.name)
        print("Salary:", self._salary)
        print("Department:", self.department)

    def calculate_bonus(self):
        return self._salary * 0.05

    # Getter
    def get_salary(self):
        return self._salary

    # Setter
    def set_salary(self, salary):
        if salary > 0:
            self._salary = salary
        else:
            print("Invalid Salary")
```

```python
# Child Class - Developer
class Developer(Employee):

    def __init__(self, emp_id, name, salary, department,
                 programming_language, projects):
        super().__init__(emp_id, name, salary, department)
        self.programming_language = programming_language
        self.projects = projects

    def display_details(self):
        super().display_details()
        print("Programming Language:", self.programming_language)
        print("Projects:", self.projects)

    def calculate_bonus(self):
        return self._salary * 0.10
```

```python
# Child Class - Manager
class Manager(Employee):

    def __init__(self, emp_id, name, salary, department,
                 team_size, management_bonus):
        super().__init__(emp_id, name, salary, department)
        self.team_size = team_size
        self.management_bonus = management_bonus

    def display_details(self):
        super().display_details()
        print("Team Size:", self.team_size)
        print("Management Bonus:", self.management_bonus)

    def calculate_bonus(self):
        return (self._salary * 0.15) + self.management_bonus
```

```python

# Child Class - HR Manager
class HRManager(Employee):

    def __init__(self, emp_id, name, salary, department,
                 employees_handled):
        super().__init__(emp_id, name, salary, department)
        self.employees_handled = employees_handled

    def display_details(self):
        super().display_details()
        print("Employees Handled:", self.employees_handled)

    def calculate_bonus(self):
        return self._salary * 0.08

```

```python
# -----------------------------
# Creating Objects
# -----------------------------

developer = Developer(
    101,
    "Ajith",
    50000,
    "IT",
    "Python",
    4
)

manager = Manager(
    102,
    "Rahul",
    80000,
    "Management",
    10,
    5000
)

hr_manager = HRManager(
    103,
    "Priya",
    60000,
    "HR",
    50
)
```

```python
# -----------------------------
# Display Details
# -----------------------------

developer.display_details()
manager.display_details()
hr_manager.display_details()
```

```python
# -----------------------------
# Encapsulation Example
# -----------------------------

print("\nSalary Before Update:", developer.get_salary())

developer.set_salary(55000)

print("Salary After Update:", developer.get_salary())

# -----------------------------
# Polymorphism Example
# -----------------------------

employees = [
    developer,
    manager,
    hr_manager
]

print("\n----- Bonus Calculation -----")

for employee in employees:
    print(
        employee.name,
        "Bonus =",
        employee.calculate_bonus()
    )

# -----------------------------
# Test Cases
# -----------------------------

print("\n----- Test Cases -----")

assert developer.calculate_bonus() == 5500.0
assert manager.calculate_bonus() == 17000.0
assert hr_manager.calculate_bonus() == 4800.0

print("All Test Cases Passed Successfully")
```
